```python
import random
import math
import time

# Generar conjuntos de datos aleatorios...

# 42 es una referencia cultural al libro "The Hitchhiker's Guide to the Galaxy".
random.seed(42)  # fijar la semilla

# Recomendado por el profesor:
# Lista de 1000 números enteros aleatorios en 1D, 2D y 3D.
LISTA_1D = [random.randrange(1, 10000) for _ in range(1000)] # datos de prueba en dimensión 1
LISTA_2D = [(random.randrange(1, 10000), random.randrange(1, 10000)) for _ in range(1000)] # datos de prueba en dimensión 2
LISTA_3D = [(random.randrange(1, 10000), random.randrange(1, 10000), random.randrange(1, 10000)) for _ in range(1000)] # datos de prueba en dimensión 3

# Funciones de distancia euclidiana entre dos puntos en 1D, 2D y 3D...
# p1 -> coordenada del primer punto.
# p2 -> coordenada del segundo punto.

# 1D -> distancia euclidiana -> valor absoluto.
def dist_1d(p1, p2):
    return abs(p1 - p2)

# 2D -> distancia euclidiana -> sqrt(Δx² + Δy²).
def dist_2d(p1, p2):
    return math.sqrt((p1[0] - p2[0])**2 + (p1[1] - p2[1])**2)

# 3D -> distancia euclidiana -> sqrt(Δx² + Δy² + Δz²).
def dist_3d(p1, p2):
    return math.sqrt((p1[0] - p2[0])**2 + (p1[1] - p2[1])**2 + (p1[2] - p2[2])**2)
```

## Solución en 1D - Fuerza Bruta

**Complejidad:**  
O(n²) -> n(n-1)/2 comparaciones.

Dado n puntos en 1D la fuerza bruta compara todos los pares posibles para encontrar el par con distancia mínima.

**Fórmula combinatoria:**  
(n 2) = n(n-1)/2

```python
def closest_1d_bruteforce(points):
    """Devuelve el par más cercano y su distancia (fuerza bruta)"""
    min_dist = float('inf')
    pair = None
    for i in range(len(points)):
        for j in range(i+1, len(points)):
            d = dist_1d(points[i], points[j])
            if d < min_dist:
                min_dist = d
                pair = (points[i], points[j])
    return pair, min_dist

# Prueba
start = time.time()
pair_bf, dist_bf = closest_1d_bruteforce(LISTA_1D)
end = time.time()

print(f"Fuerza Bruta 1D -> Par: {pair_bf}, Distancia: {dist_bf:.6f}")
print(f"Tiempo: {end-start:.4f} segundos\n")
```

### Complejidad...

- Bucle externo -> **n** iteraciones.
- Bucle interno -> en promedio **~n/2** iteraciones.
- Total comparaciones -> exactamente **n(n-1)/2**  
  
**Complejidad = O(n²)**  
**¿Se puede mejorar?**  
Sí, con Divide y Vencerás: O(n log n).

## Solución en 1D - Divide y Vencerás.

**Idea:** Ordenar + recursión. Solo es necesario comparar puntos consecutivos después de ordenar.

```python
def closest_1d_dyv(points):
    """Algoritmo Divide y Vencerás (DyV) para encontrar los dos puntos más cercanos"""
    # Una vez ordenados los puntos, los candidatos más cercanos siempre están consecutivos o muy cerca en la lista ordenada.
    points = sorted(points) # ordenación inicial

    def dyv(left, right):
        if right - left <= 3:  # caso base (≤ 3 puntos) -> fuerza bruta
            min_d = float('inf')
            best_pair = None
            for i in range(left, right):
                for j in range(i+1, right):
                    d = dist_1d(points[i], points[j])
                    if d < min_d:
                        min_d = d
                        best_pair = (points[i], points[j])
            return min_d, best_pair

        mid = (left + right) // 2 # divide el intervalo [left, right) en dos mitades casi iguales
        d1, pair1 = dyv(left, mid) # obtiene distancia mínima d1
        d2, pair2 = dyv(mid, right) # obtiene distancia mínima d2

        delta = min(d1, d2) # mejor distancia hasta ahora
        best_pair = pair1 if d1 < d2 else pair2 # mejor par (con la distancia más pequeña) hasta ahora

        # Los dos puntos inmediatamente adyacentes al punto medio.
        if dist_1d(points[mid-1], points[mid]) < delta:
            delta = dist_1d(points[mid-1], points[mid])
            best_pair = (points[mid-1], points[mid])

        return delta, best_pair # mejor distancia y el par correspondiente

    return dyv(0, len(points))

# Prueba
start = time.time()
pair_dyv, dist_dyv = closest_1d_dyv(LISTA_1D.copy())
end = time.time()

# pair_dyv = distancia (float)
# dist_dyv = par (tupla)

print(f"Divide y Vencerás 1D -> Par: {dist_dyv}")
print(f"           Distancia: {pair_dyv:.6f}")
print(f"Tiempo: {end - start:.6f} segundos (mucho más rápido)\n")
```

### Complejidad...

- Ordenación inicial: O(n log n).
- Recurrencia: T(n) = 2 T(n/2) + O(1) -> O(n).
- Total -> O(n log n) (domina la ordenación).
  
**¿Se puede mejorar?**  
No, no en términos de complejidad (la O(n log n) es óptima).

## Extensión a 2D - Divide y Vencerás.

Algoritmo clásico de 1977 (Shamos-Hoey). Complejidad óptima O(n log n).

```python
def closest_2d(points):
    if len(points) <= 1:
        return None, float('inf')

    px = sorted(points, key=lambda p: p[0])  # ordenado por x
    py = sorted(points, key=lambda p: p[1])  # ordenado por y

    def recurse(px, py):
        n = len(px)
        if n <= 3:
            # Fuerza bruta para casos pequeños.
            min_d = float('inf')
            best = None
            for i in range(n):
                for j in range(i+1, n):
                    d = dist_2d(px[i], px[j])
                    if d < min_d:
                        min_d = d
                        best = (px[i], px[j])
            return min_d, best

        mid = n // 2
        mid_x = px[mid][0]

        pyl = [p for p in py if p[0] <= mid_x]
        pyr = [p for p in py if p[0] > mid_x]

        dl, pair_l = recurse(px[:mid], pyl)
        dr, pair_r = recurse(px[mid:], pyr)

        delta = min(dl, dr)
        best_pair = pair_l if dl < dr else pair_r

        # Tira central de ancho 2*delta.
        strip = [p for p in py if abs(p[0] - mid_x) < delta]

        for i in range(len(strip)):
            for j in range(i+1, min(i+8, len(strip))):  # máximo 7 puntos
                if strip[j][1] - strip[i][1] >= delta:
                    break
                d = dist_2d(strip[i], strip[j])
                if d < delta:
                    delta = d
                    best_pair = (strip[i], strip[j])

        return delta, best_pair

    return recurse(px, py)

# Prueba
start = time.time()
pair_2d, dist_2d = closest_2d(LISTA_2D.copy())
end = time.time()

print(f"Divide y Vencerás 2D -> Par: {dist_2d}")
print(f"           Distancia: {pair_2d:.6f}")
print(f"Tiempo: {end - start:.6f} segundos\n")
```

## Extensión a 3D - Divide y Vencerás

Complejidad teórica peor caso O(n²), pero en la práctica con datos aleatorios es rápido: O(n log n).

```python
def closest_3d(points):
    if len(points) < 2:
        return None, float('inf')

    px = sorted(points, key=lambda p: p[0])

    def recurse(px):
        n = len(px)
        if n <= 3:
            min_d = float('inf')
            best = None
            for i in range(n):
                for j in range(i+1, n):
                    d = dist_3d(px[i], px[j])
                    if d < min_d:
                        min_d = d
                        best = (px[i], px[j])
            return min_d, best

        mid = n // 2
        mid_x = px[mid][0]

        dl, pair_l = recurse(px[:mid])
        dr, pair_r = recurse(px[mid:])

        delta = min(dl, dr)
        best_pair = pair_l if dl < dr else pair_r

        # Tira/losa.
        strip = [p for p in px if abs(p[0] - mid_x) < delta]

        for i in range(len(strip)):
            for j in range(i+1, len(strip)):
                if strip[j][1] - strip[i][1] >= delta:
                    break
                d = dist_3d(strip[i], strip[j])
                if d < delta:
                    delta = d
                    best_pair = (strip[i], strip[j])

        return delta, best_pair

    return recurse(px)

# Prueba
start = time.time()
pair_3d, dist_3d = closest_3d(LISTA_3D.copy())
end = time.time()

print(f"Divide y Vencerás 3D -> Par: {dist_3d}")
print(f"           Distancia: {pair_3d:.6f}")
print(f"Tiempo: {end - start:.6f} segundos\n")
```

## Conclusiones:

| Dimensión | Fuerza Bruta | Divide y Vencerás | Complejidad Óptima |
|-----------|--------------|-------------------|---------------------|
| 1D        | O(n²)        | O(n log n)        | O(n log n)          |
| 2D        | O(n²)        | O(n log n)        | O(n log n)          |
| 3D        | O(n²)        | O(n log n) práctico| O(n log² n) teórico |
